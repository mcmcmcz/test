```python
import pandas as pd
import numpy as np
import os
import json
```

```python
base_dir = '../writing_agents_results/stage_metrics_1/'
files = [file for file in os.listdir(base_dir) if file.endswith('.json')]
all_data = []
for file in files:
    with open(base_dir + file, 'r', encoding='utf-8') as fp:
        data = json.load(fp)
    all_data.append(data)
```

```python
print(np.mean([d['fluency'] for d in all_data]))
print(np.mean([d['consistency'] for d in all_data]))
print(np.mean([d['logic'] for d in all_data]))
```
